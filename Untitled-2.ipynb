## Grouping Age into Categories
Age is a continuous value, so I bucketed it into ranges (Child, Teen, Young Adult, Adult, Senior) to make it easier to group and compare.

```python
df['AgeGroup'] = pd.cut(df['Age'], bins=[0, 12, 18, 35, 60, 100], labels=['Child', 'Teen', 'Young Adult', 'Adult', 'Senior'])
df[['Age', 'AgeGroup']].head(10)
```

## Key Findings
- Sex was the strongest predictor of survival — women survived at much higher rates than men across almost every class and age group (e.g. ~97% for 1st class women vs. ~37% for 1st class men)
- Class mattered too, especially for men — 1st class men survived more than 3x as often as 2nd or 3rd class men
- One result (1st class female children at 0% survival) turned out to be based on a single passenger, not a real trend — a reminder to check group size before trusting a percentage
